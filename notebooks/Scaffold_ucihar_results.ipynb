```python
import pandas as pd

df = pd.read_csv("../results/uci_scaffold_r50.csv")  
df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["round"], df["accuracy"])
plt.xlabel("Communication Round")
plt.ylabel("Accuracy")
plt.title("SCAFFOLD on UCI-HAR (α = 0.1)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
