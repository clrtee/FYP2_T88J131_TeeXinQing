```python
import pandas as pd

df = pd.read_csv("../results/casas_fedprox_r50.csv") 
df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["round"], df["accuracy"])
plt.xlabel("Communication Round")
plt.ylabel("Accuracy")
plt.title("FedProx on CASAS (α = 0.1)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure()
plt.plot(df["round"], df["f1"])
plt.xlabel("Communication Round")
plt.ylabel("F1-score")
plt.title("FedProx on CASAS (α = 0.1)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
